```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
base = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/{}/results.csv"


df_2022 = pd.read_csv(base.format(2022))
df_2023 = pd.read_csv(base.format(2023))
df_2024 = pd.read_csv(base.format(2024))
df_2025 = pd.read_csv(base.format(2025))


#with pd.option_context("display.max_rows", None):
      #print(df_2025.isnull().sum())
#print(df_2024.shape)
#df_2025.head()
```

```python
cols_2022 = set(df_2022.columns)
cols_2023 = set(df_2023.columns)
cols_2024 = set(df_2024.columns)
cols_2025 = set(df_2025.columns)

common_cols = (cols_2022) & (cols_2023) & (cols_2024) & (cols_2025)

cols = sorted(common_cols)
```

```python
small_2022 = df_2022[cols].copy()
small_2022["Year"] = 2022

small_2023 = df_2023[cols].copy()
small_2023["Year"] = 2023

small_2024 = df_2024[cols].copy()
small_2024["Year"] = 2024

small_2025 = df_2025[cols].copy()
small_2025["Year"] = 2025



df = pd.concat([small_2022, small_2023, small_2024, small_2025], ignore_index = True)

print(df[df['MainBranch'] == 'I am learning to code'].isnull().sum())

print(df.info())


```

```python
# Data base uniqe

array_db=df['DatabaseHaveWorkedWith'].unique()

unique_db_hw = (
    pd.Series(array_db)
    .dropna()
    .str.split(';')
    .explode()
    .str.strip()
    .unique()
)

print(unique_db_hw)
```
