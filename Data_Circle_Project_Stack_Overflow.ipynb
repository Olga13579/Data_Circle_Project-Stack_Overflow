```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url1 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2025/results.csv"
url2 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2024/results.csv"
url3 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2023/results.csv"
url4 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2022/results.csv"

df25 = pd.read_csv(url1)
df24 = pd.read_csv(url2)
df23 = pd.read_csv(url3)
df22 = pd.read_csv(url4)

print(df25.shape)
print(df24.shape)
print(df23.shape)
print(df22.shape)
```

```python
url5 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2025/schema.csv"
url6 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2024/schema.csv"
url7 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2023/schema.csv"
url8 = "https://github.com/StackExchange/Survey/raw/refs/heads/main/packages/archive/2022/schema.csv"

sdf25 = pd.read_csv(url5)
sdf24 = pd.read_csv(url6)
sdf23 = pd.read_csv(url7)
sdf22 = pd.read_csv(url8)

print(sdf25.shape)
print(sdf24.shape)
print(sdf23.shape)
print(sdf22.shape)
```

```python
print(sdf25.columns)
print(sdf24.columns)
print(sdf23.columns)
print(sdf22.columns)
```

```python
cols25_s = set(sdf25["qname"])
cols24_s = set(sdf24["qname"])
cols23_s = set(sdf23["qname"])
cols22_s = set(sdf22["qname"])

#the same questions
common_all = cols25_s & cols24_s & cols23_s & cols22_s
display(len(common_all), common_all)
```

```python
def find_columns(df, qname):
    return [col for col in df.columns if col.lower().startswith(qname.lower())]

for q in list(common_all):
    print(f"\n=== {q} ===")

    print("2025:", find_columns(df25, q))
    print("2024:", find_columns(df24, q))
    print("2023:", find_columns(df23, q))
    print("2022:", find_columns(df22, q))
```

```python
cols25 = []
for q in common_all:
    cols25.extend(find_columns(df25, q))

cols24 = []
for q in common_all:
    cols24.extend(find_columns(df24, q))

cols23 = []
for q in common_all:
    cols23.extend(find_columns(df23, q))

cols22 = []
for q in common_all:
    cols22.extend(find_columns(df22, q))

display(df25[cols25].head(3))
display(df24[cols24].head(3))
display(df23[cols23].head(3))
display(df22[cols22].head(3))
```
