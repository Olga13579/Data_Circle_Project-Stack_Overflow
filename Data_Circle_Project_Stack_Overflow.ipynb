```python
datasets = {
    "2025": df25,
    "2024": df24,
    "2023": df23,
    "2022": df22
}

for year, df in datasets.items():
    total = len(df)
    non_null = df["ConvertedCompYearly"].notna().sum()
    print(f"{year}: non_null {non_null} from total {total}")
```

```python
#Removing rows with missing salary values (ConvertedCompYearly)
df22_clean = df22[df22["ConvertedCompYearly"].notna()].copy()
df23_clean = df23[df23["ConvertedCompYearly"].notna()].copy()
df24_clean = df24[df24["ConvertedCompYearly"].notna()].copy()
df25_clean = df25[df25["ConvertedCompYearly"].notna()].copy()

#Adding a new column "Year" indicating the survey year
df22_clean["Year"] = 2022
df23_clean["Year"] = 2023
df24_clean["Year"] = 2024
df25_clean["Year"] = 2025

#Selecting common columns and creating a big common database
# 1. Collect qname sets for each year
cols25_s = set(sdf25["qname"])
cols24_s = set(sdf24["qname"])
cols23_s = set(sdf23["qname"])
cols22_s = set(sdf22["qname"])

# 2. Find qnames that exist in all four years
common_qnames = cols25_s & cols24_s & cols23_s & cols22_s
print(f"Common qnames count: {len(common_qnames)}")

# 3. Function to find real column names by qname prefix
def find_columns(df, qname):
    return [col for col in df.columns if col.lower().startswith(qname.lower())]

# 4. Expand qnames into actual column names for each dataset
cols22 = []
cols23 = []
cols24 = []
cols25 = []

for q in common_qnames:
    cols22.extend(find_columns(df22_clean, q))
    cols23.extend(find_columns(df23_clean, q))
    cols24.extend(find_columns(df24_clean, q))
    cols25.extend(find_columns(df25_clean, q))

# 5. Add Year and ConvertedCompYearly manually (they are not part of qname)
cols22 += ["Year", "ConvertedCompYearly"]
cols23 += ["Year", "ConvertedCompYearly"]
cols24 += ["Year", "ConvertedCompYearly"]
cols25 += ["Year", "ConvertedCompYearly"]

# 6. Select only these columns
df22_sel = df22_clean[cols22]
df23_sel = df23_clean[cols23]
df24_sel = df24_clean[cols24]
df25_sel = df25_clean[cols25]

# 7. Combine all years into one unified dataset
df_all = pd.concat([df22_sel, df23_sel, df24_sel, df25_sel], ignore_index=True)

df_all.shape
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df_all["ConvertedCompYearly"], bins=100, color='skyblue', edgecolor='black')
plt.title("Overall salary distribution (all years)")
plt.xlabel("Salary (USD)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(np.log1p(df_all["ConvertedCompYearly"]), bins=100, color='salmon', edgecolor='black')
plt.title("Overall salary distribution (log scale)")
plt.xlabel("log(1 + salary)")
plt.ylabel("Count")
plt.show()
```

```python
#Basic statistics
print("Overall salary distribution")
print(df_all["ConvertedCompYearly"].describe())

#Check suspiciously high/low and extremely high/low salaries
print("Salary > 500K USD", (df_all["ConvertedCompYearly"] > 500000).sum())
print("Salary > 1M USD)", (df_all["ConvertedCompYearly"] > 1000000).sum())
print("Salary < 100 USD:", (df_all["ConvertedCompYearly"] < 100).sum())
print("Salary < 500 USD:", (df_all["ConvertedCompYearly"] < 500).sum())
```

```python
#Delete rows with unreal salary values
df_all_clean = df_all[(df_all["ConvertedCompYearly"] >= 500) & (df_all["ConvertedCompYearly"] <= 500000)]
print(df_all_clean.shape)
```
